```python
import numpy as np
import matplotlib.pyplot as plt

# ==========================================
# ГРАФІК 1: Вольт-амперна характеристика (ВАХ)
# ==========================================

# Дані для Варіанту 1 (Ф1 = 2.50 лм)
U1 = np.array([4, 8, 12, 16, 20, 26, 32, 36, 40, 43, 46, 50])
I1 = np.array([20, 30, 40, 50, 60, 70, 76, 78, 80, 80, 80, 80])

# Дані для Варіанту 2 (Ф2 = 1.98 лм)
U2 = np.array([6, 10, 14, 20, 28, 30, 33])
I2 = np.array([20, 30, 40, 50, 56, 56, 56])

plt.figure(figsize=(10, 6))

# Будуємо криві ВАХ (з маркерами точок)
plt.plot(U1, I1, marker='o', linestyle='-', color='blue', linewidth=2, markersize=6, label=r'$\Phi_1 = 2.50$ лм (Варіант 1)')
plt.plot(U2, I2, marker='s', linestyle='-', color='red', linewidth=2, markersize=6, label=r'$\Phi_2 = 1.98$ лм (Варіант 2)')

# Пунктирні лінії для показу струму насичення
plt.axhline(80, color='blue', linestyle='--', alpha=0.5)
plt.axhline(56, color='red', linestyle='--', alpha=0.5)
plt.text(2, 82, r'$I_{нас1} = 80$ мкА', color='blue', fontsize=10)
plt.text(2, 58, r'$I_{нас2} = 56$ мкА', color='red', fontsize=10)

# Оформлення першого графіка
plt.title('Вольт-амперна характеристика (ВАХ) фотоелементу', fontsize=14, pad=15)
plt.xlabel('Прикладена напруга $U$, В', fontsize=12)
plt.ylabel('Сила фотоструму $I$, $10^{-6}$ А (мкА)', fontsize=12)
plt.grid(True, linestyle=':', alpha=0.7)
plt.legend(loc='lower right', fontsize=11, framealpha=1, edgecolor='black')
plt.xlim(0, 55)
plt.ylim(0, 90)

# ==========================================
# ГРАФІК 2: Світлова характеристика
# ==========================================

# Дані світлового потоку (спільні для обох напруг)
Phi = np.array([0.18, 0.20, 0.24, 0.28, 0.33, 0.40, 0.49, 0.63, 0.82, 1.11, 1.60, 2.50])

# Дані струму для U = 40 В
I_40 = np.array([4, 6, 7, 8, 10, 12, 14, 20, 22, 30, 46, 78])

# Дані струму для U = 50 В
I_50 = np.array([4, 6, 7, 8, 10, 12, 14, 18, 24, 32, 46, 74])

plt.figure(figsize=(10, 6))

# Будуємо ТІЛЬКИ точки (без з'єднувальних ліній)
plt.scatter(Phi, I_40, color='teal', marker='o', s=40, label='Експеримент $U = 40$ В', zorder=5)
plt.scatter(Phi, I_50, color='darkorange', marker='s', s=40, label='Експеримент $U = 50$ В', zorder=5)

# Додаємо ідеальні прямі (лінії тренду) з нуля
m_40, b_40 = np.polyfit(Phi, I_40, 1)
m_50, b_50 = np.polyfit(Phi, I_50, 1)

x_line = np.linspace(0, 2.6, 100)
plt.plot(x_line, m_40*x_line, color='teal', linestyle='-', linewidth=2, label='Лінія тренду ($U=40$ В)')
plt.plot(x_line, m_50*x_line, color='darkorange', linestyle='-', linewidth=2, label='Лінія тренду ($U=50$ В)')

# Оформлення другого графіка
plt.title('Світлова характеристика фотоелементу', fontsize=14, pad=15)
plt.xlabel(r'Світловий потік $\Phi$, лм', fontsize=12)
plt.ylabel('Сила фотоструму $I$, $10^{-6}$ А (мкА)', fontsize=12)
plt.grid(True, linestyle=':', alpha=0.7)
plt.legend(loc='upper left', fontsize=11, framealpha=1, edgecolor='black')
plt.xlim(0, 2.7)
plt.ylim(0, 85)

plt.show()
```
