Застосуйте Lasso і порівняйте результат зі звичайною лінійною регресією:
Згенеруйте дані за допомогою make_regression (n_samples=500, n_features=20, n_informative=5, noise=10, random_state=42).

* Навчіть модель LinearRegression і модель Lasso(alpha=0.1).
* Розділіть дані на train/test(80/20, random_state=42).
* Порівняйте R² на тестовій вибірці.
* Виведіть кількість коефіцієнтів, які Lasso занулила.

```python
from sklearn.model_selection import train_test_split
from sklearn.datasets import make_regression
from sklearn.linear_model import LinearRegression, Lasso, Ridge
X3, y3 = make_regression(n_samples=500, n_features=20, n_informative=5, noise=10, random_state=42)
```

```python
xtrain, xtest, ytrain, ytest = train_test_split(
    X3,
    y3,
    test_size = 0.2,
    random_state = 42,
)
```

```python
lr_model = LinearRegression()
```

```python
lasso_model = Lasso(alpha=0.1)
lr_model.fit(xtrain, ytrain)
lasso_model.fit(xtrain, ytrain)
print(*(f'|{c1:.3f} |\t| {c2:.3f}|' for c1, c2 in zip(lr_model.coef_, lasso_model.coef_)), sep='\n')
```

```python
import sklearn.metrics as sm
lr_predict = lr_model.predict(xtest)
lasso_predict = lasso_model.predict(xtest)
print(f'Coefecient of determination: linear regression: {sm.r2_score(ytest, lr_predict)}, LASSO-model: {sm.r2_score(ytest, lasso_predict)}')
```

```python
import numpy as np
num_zero_coefs_lasso = np.sum(lasso_model.coef_ == 0)
num_zero_coefs_lr = np.sum(lr_model.coef_ == 0)
print(f'Quantity of coefficients, which Lasso lost: {num_zero_coefs}')
```

```python
ridge_model = Ridge(alpha=1.0)
ridge_model.fit(xtrain, ytrain)
num_zero_coefs_ridge = np.sum(ridge_model.coef_ == 0)
```

```python
print(f'Determetaion coeficient (R^2) for Ridge model: {sm.r2_score(ytest, ridge_model.predict(xtest))}')
```
